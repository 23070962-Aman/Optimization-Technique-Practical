```python
import sympy as sp
```

```python
# variable
x,lam  = sp.symbols('x lam', nonnegative=True)

```

```python
# objective function
f = x**2 + x + 1
```

```python
# constraint
# x>=2
g = 2-x
```

```python
# langrangian
L = f + lam * g
```

```python
#Stationary condition
eq1 = sp.diff(L,x)
```

```python
# Complementary slackness
eq2 = lam*g
```

```python

#Solving equation
solutions = sp.solve((eq1,eq2),(x,lam), dict=True)
print("Possible solution")
print(solutions)
```

```python
for sol in solutions:
  x_val = sol[x]
  lam_val = sol[lam]

  if x_val>=2 and lam_val >= 0:
    print("\nOptimsl Solution")
    print("x =",x_val)
    print("Lambda =",lam_val)
    print("Minimum Value =",f.subs(x,x_val))
```
